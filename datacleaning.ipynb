```python
import pandas as pd
df=pd.read_csv("Data_clean.csv");
```

```python
df.head()
```

```python
#missing values code here
df["Name"].isnull().unique() #name has no null
df["Age"].isnull().unique() #age has no null
df["City"].isnull().unique() #city has no null
df["Gender"].isnull().unique() #gender has no null
df["Stream"].isnull().unique() #stream has no null
df["Marks"].isnull().unique() #marks has null
df["Marks"].value_counts().unique()
avg_mean=df["Marks"].mean()
df["Marks"]=df["Marks"].fillna(avg_mean)
```

```python
df["Marks"].value_counts().unique()
df["Marks"].isnull().unique()
```

```python
df["StudyHours"].isnull().unique() #study hours has no null
```

```python
df["Attendance"].isnull().unique() #Attendance has no null or nan
df["Passed"].isnull().unique() #passed has no null or nan
```

```python
df.isnull().sum() #the one whole code to check which column has null
```

```python
#now check duplicates for whole column
df.duplicated().sum()
#df.drop_duplicates() this checks if entire row corresponds to another row and remove
df=df.drop_duplicates(subset="StudentID") #this checks if studentId is same and remove
```

```python
#using loc to update the city values, other values remain constant
df.loc[:,"City"]=df["City"].str.capitalize().replace({
    "Ktm":"Kathmandu",
})
```

```python
df["Gender"].unique()
```

```python
df.loc[:,"Gender"]=df["Gender"].str.capitalize().replace({
    "F":"Female",
    "M":"Male",
    "Femal":"Female"
})
```

```python
df["Stream"].unique()
```

```python
df.loc[:,"Stream"]=df["Stream"].str.capitalize().replace({
    "Com":"Commerce",
    "Sci":"Science",
})
```

```python
df["Stream"].unique()
```

```python
df["Passed"].unique()
```

```python
df.loc[:,"Passed"]=df["Passed"].str.capitalize().replace({
    "Y":"Yes",
    "N":"No"
})
```
